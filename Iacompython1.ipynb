```python

#minha resolucao do jogo
import random

# Geração Aleatória: sorteia um número entre 1 e 5
numero_secreto = random.randint(1, 5)
total_tentativas = 3

# Interatividade: Informando as regras
print("⚔️  BEM-VINDO AO DUELO CONTRA A IA ⚔️") #tipo uma introducao
print("A IA escolheu um número entre 1 e 5.") #tipo uma introducao
print(f"Você tem {total_tentativas} chances para acertar. Boa sorte!\n") ##tipo uma introducao, fala quantas tentativas voce tem para acertar o numero sorteado

# Sistema de Tentativas: Laço de repetição 'for'
for tentativa in range(1, total_tentativas + 1): # 1 onde a contagem comeca
    # Input: O jogador faz o seu palpite
    palpite = int(input(f"Tentativa {tentativa} de {total_tentativas} - Qual seu palpite? ")) # Serve para o jogador saber exatamente em qual tentativa está.

    # Estrutura Condicional: Verificando o palpite
    if palpite == numero_secreto:
        print("✨ VITÓRIA! Você desvendou o código da IA!") # aqui se voce acertar o numero ele mostra esse texto
        # Interrupção de Fluxo: Encerra o laço imediatamente
        break
    else:
        if tentativa < total_tentativas: # total_tentativas é a qtd que voice pode chutar o erro
            print("Errou! Tente novamente.")
        else:
            print("Suas chances acabaram!")

# Finalização: Revela o número secreto ao final (ganhando ou perdendo)
print("-" * 40)
print(f"FIM DE JOGO! O número secreto da IA era: {numero_secreto}") # Ele permite que você coloque variáveis diretamente dentro do texto.
print("-" * 40) #: deixar a mensagem de "FIM DE JOGO" centralizada entre duas linhas iguais
```
