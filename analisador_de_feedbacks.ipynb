```python
#1 importação das bibliotecas

import spacy # biblioteca p/ texto em numero
import nltk # biblioteca p/  probabilidade para prever o sentimento da pessoa
from nltk.sentiment.vader import SentimentIntensityAnalyzer
```

```python
# 2 Preparação

!python -m spacy download pt_core_news_sm
nltk.download('vader_lexicon')


nlp = spacy.load("pt_core_news_sm")
termometro_lntk = SentimentIntensityAnalyzer()
```

```python
# 3 Escala de sentimentos (Para a NLTK não dar nota zero no Português) aqui que faz as curadorias
pesos_extras = {
    "maravilhoso" : 1.0, "adorar": 2.0, "incrivel" : 4.0,
    "péssimo" : -0.5, "ruim" : -0.4, "confuso" : -0.4
}

#Atualizar o dicionario interno do nltk vom as palavras em Português
termometro_lntk.lexicon.update(pesos_extras)
```

```python
# 4 Dados

frases_clientes = [
    "O curso do Senai é maravilhoso, os professores são muito atenciosos!",
    "Péssimo atendimento. O suporte demorou demais para responder.",
    "A aula de hoje foi normal, nada de muito especial.",
    "EU ADOREI o conteúdo de IA! É muito fácil de entender.",
    "Não gostei do material, achei confuso e muito caro.",
    "O evento em São Paulo foi incrível, mas o café estava ruim."
]
```

```python
#  5 Execucaco Integrada

print(f"{'FEEDBACKKKKK ORIGINAL':<70} | {'NOTA':<6} | {'SENTIMENTO'}") #<70 é o espaçamento
print("_" * 90)

for texto in frases_clientes:
    #Passo A: spacy limpa o texto
    doc = nlp(texto.lower()) # convertendo para minusculo
    palavras_uteis = {token.lemma_ for token in doc if not token.is_stop and not token.is_punct}
    frase_limpa = " ".join(palavras_uteis)

    #Passo B: NLTK mede o sentimento
    #NLTK recebe o texto limpo pelo spacy

    nota = termometro_lntk.polarity_scores(frase_limpa)['compound']

    if nota >= 0.1:
        resultado = "😊😊 POSITIVO"
    elif nota <= -0.1:
        resultado = "🤬🤬 NEGATIVO"
    else:
        resultado = "🤨🤨 NEUTRO"

    print(f"{texto:<70} | {nota:<6.1f} | {resultado}")
```
