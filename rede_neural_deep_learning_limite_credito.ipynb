```python
#Célula 1 : Importação das bibliotecas

import pandas as pd
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# Célula 2 : Carregar os dados
df = pd.read_csv("limites_de_credito.csv")

X = df[['Idade','Salario','Score_Credito']]
y = df[['Limite_Aprovado']]
```

```python
#Célula 3: Normalizador X (Entradas) e tamebm y (Alvo)
normalizador_x = StandardScaler()
X_norm = normalizador_x.fit_transform(X)

normalizdor_y = StandardScaler()
#O StandardSclaer exige o formato 2D, por isso usamos o .values.resshape(-1,1)
y_norm = normalizdor_y.fit_transform(y.values.reshape(-1, 1))

#Vamos passar as duas variaveis normalizadas para o treino
X_treino, X_teste, y_treino, y_teste = train_test_split(X_norm, y_norm, test_size=0.2, random_state=42)
```

```python
#Célula 4 : Construindo a inteligência Artificial

modelo_limite = keras.Sequential([
keras.Input(shape=(3,)), # 3 em especifico porque nessa situacao sao 3 informacoes descrita la em cima
keras.layers.Dense(8,activation='relu'), # relu
keras.layers.Dense(4,activation='relu'),
keras.layers.Dense(1) #Devolve um numero puro (mas ainda normalizado)

])

modelo_limite.compile(optimizer='adam', loss='mse', metrics=['mae'])
```

```python
#Célula 5: Treinar o modelo

print("Treina o avaliador de limites...")
modelo_limite.fit(X_treino, y_treino, epochs=100, verbose=0) # para nao gerar ate o 100 por o verbose
print("Treino Concluído!\n")
```

```python
# Célula 6 : Teste com Cliente
# Cliente tem 35 anos, salário de R$ 8.000 e score de 750

cliente_novo = pd.DataFrame({'Idade': [35], 'Salario': [15000], 'Score_Credito': [1000]})
cliente_norm = normalizador_x.transform(cliente_novo)

# A IA predizer um valor pequeno (ex: 0.15)
limite_previsto_norm = modelo_limite.predict(cliente_norm)

#Usamos o tradutor para reveter o valor para reais l
limite_real = normalizdor_y.inverse_transform(limite_previsto_norm)

print(f"O limite de crédito sugerido é: {limite_real[0][0]:.2f}")
```
