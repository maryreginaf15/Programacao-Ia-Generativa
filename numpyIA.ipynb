```python
# A matematica da IA(Numpy)

import numpy as np # as serve para nomear o nome da biblioteca

print("-----O PROBLEMA DO PYTHON PURO----------")
lista_preco = [1000,200000,30000]

#O python puro nao faz a conta apenas duplica os elementos da lista!

print(f"Lista vezes 2:{lista_preco*2}")

print("\n ----   A SOLUÇÃO COM NUMPY ")
#Trasformamos a lista num 'array (um super vetor matematico)
precos_numpy = np.array([100000,200000,3000])

# O NumPy perceve instaneamente que queremos multiplicar cada numero
precos_atualizados = precos_numpy * 2
print(f"Array vezes 2 {precos_atualizados}")

print("\n----- Como a IA VÊ O MUNDO (Matrizes)---- ")

#Na IA (especialmente em IA), tudo sao numeros em matrizes
matriz_dados = np.array([
[1 ,2 ,3],
[4, 5 ,6]

])
print(matriz_dados)
```
