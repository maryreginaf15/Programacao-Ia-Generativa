```python
# Celula 1 : Preparando o terreno

import pandas as pd

from sklearn.linear_model import LinearRegression

#criando os nossos dados historicos (o passado para ensinar a maquina)

dados = { # usando dicionario
'Tamanho_m2' :  [50,70,80,100,120], # usando lista com valores
          'preco': [150000,210000,240000,300000,360000]
}

df = pd.DataFrame(dados)
print("Os nossos dados de treino:")
display(df)
```

```python
# Celula 2 : Gerar um grafico de dispersao (Scatter Plot)

df.plot(kind='scatter', x='Tamanho_m2', y='preco', color='purple', title='Relação: Tamanho x Preço')
```

```python
# Celula 3 : Separando X e y
# o X é maisculo é pq é uma matriz

X = df[['Tamanho_m2']] # 2 [] lista dentro de lista ou seja uma tabela dentro de lista
y = df['preco']

```

```python
# Celula 4 :  criando e treinando a IA

modelo_ia = LinearRegression() #criando um "cerebro" vazio

#vamos treinar a I Passando os tamanhos(X) e os precos(y)
modelo_ia.fit(X,y)

print("A inteligencia artificial foi treinada com sucesso!")
```

```python
# Celula 5: A hora da verdade(Predição)
# vamos ciar uma mini-tabela com 1 linha para a nova casa, como mesmo nome da coluna

tamanho_casa_nova = pd.DataFrame({'Tamanho_m2': [2000]}) #queremos saber o precos de uma casa de 90m²

#Utilizaremos o metodo .preditc()para a maquina predizer o preco

preco_estimado =modelo_ia.predict(tamanho_casa_nova) # modelos_ia sao os dados que colocou la em cima

print(f"A IA avaliada  a casa em R$: {preco_estimado[0]: .2f}") # 2f é usado para float
```

```python
#celula 6 : os dados e graficos

import pandas as pd


# dados historicos de venda da loja
dados_vendas_sorvetes = {
'temperaturas_c': [20,22,25,28,30,35],
    'sorvetes_vendidos': [50,65,80,100,120,180]

}
df_sorvetes = pd.DataFrame(dados_vendas_sorvetes)
print("Nosso historico de vendas:")
display(df_sorvetes)

#grafico para visualizacao a correlacao

df_sorvetes.plot(kind='scatter', x='temperaturas_c', y='sorvetes_vendidos', color='red', title='Temperatura vs Sorvetes Vendidos')
#
```

```python
from IPython.utils.io import temp_pyfile
# treino e predicao da IA

from sklearn.linear_model import LinearRegression

#1. Separar Pergunta (x) e resposta (y)

X = df_sorvetes[['temperaturas_c']]
y = df_sorvetes['sorvetes_vendidos']

#2. Treinar o modelo

modelo_sorvetes = LinearRegression()
modelo_sorvetes.fit(X,y)

# 3. Predicao: Amanha vai fazer uma onda de calor de 38 graus
temperatura_amanha = pd.DataFrame({'temperaturas_c': [26]})
predicao = modelo_sorvetes.predict(temperatura_amanha)

print(f"ALERTA DE ESOQUE : a predição de vendas para amanhã é {int(predicao[0])} sorvete " ) # aqui a predicao [0] sempre vai ser 0 se for outro numero da erro pois sempre pega do indice
```

```python
# Dados para a dinâmica
import pandas as pd

# Histórico dos meses anteriores
dados_marketing = {
    'Investimento_Anuncios': [1000, 2000, 3000, 4000, 5000],
    'Faturamento': [15000, 25000, 36000, 48000, 58000]
}

tabela_mkt = pd.DataFrame(dados_marketing)
display(tabela_mkt)
```

```python
# 3 Gráfico de dispersão

tabela_mkt.plot(kind='scatter', x='Investimento_Anuncios', y='Faturamento', title='Faturamento x Investimento Anúncios')

#separa o x do y
X = tabela_mkt[['Investimento_Anuncios']]
y = tabela_mkt['Faturamento']


#treino o modelo
modelo = LinearRegression()
modelo.fit(X, y)
```

```python

#predicao se a empresa investir 8000 o que acontece?
# Pergunta do diretor: o que acontece se investirmos 8000?

investimento_diretor = pd.DataFrame({'Investimento_Anuncios': [8000]})
predicao = modelo.predict(investimento_diretor)

print(f"Se investirmos 8000, o faturamento previsto de será: {predicao[0]}")
```
