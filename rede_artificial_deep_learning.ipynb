```python
#! celula : carergando os dados do banco

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

#1 ler o arquivo csv
tabela_banco = pd.read_csv('historico_credito_100_clientes.csv')

print("Visualizando os primeiros 5 clientes da base de dados : \n")
display(tabela_banco.head())
```

```python
#celula 2 : separando X (Caracteristicas ) e y (Alvo)

X = tabela_banco[['Idade', 'Salario', 'Divida_Atual']]
y = tabela_banco ['Risco_Alto']
```

```python
# CELULA 3 : Normalizando os dados (Obrigatorio para rede neurais Artificiais)

normalizado = StandardScaler()
X_normalizado = normalizado.fit_transform(X)
```

```python
# 4 Separando Treino e Teste

X_treino, X_teste, y_treino, y_teste = train_test_split(X_normalizado, y, test_size=0.2, random_state=42)

print("\nBase com 100 clientes pronta e normalizada para a rede Neural!")
```

```python
# 5 Contruindo a arquitetura de rede neural

import tensorflow as tf
from tensorflow import keras

#1 Iniciando a rede neural sequencial (camadas em sequencias )
modelo_rn = keras.Sequential([
#temos 3 informacoes de entrada: idade, salario e divida atual
    keras.Input(shape=(3,)),

  #1° Camada Oculta e quantidade de neuronios
    keras.layers.Dense(1,activation='relu'),

    #Camada saida
    keras.layers.Dense(6,activation='sigmoid')
])

#2 Compilando o Cérebro (Definindo as regras de estudo)
modelo_rn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

print("Cérebro digital contruido com sucesso! veja o resumo abaixo:")
modelo_rn.summary()
```

```python
# 6 Célula : O treinamento (Deixando a maquina estudar )

# Vamos passar os dados de treino e fazer a maquina treinar tudo 50 vezes

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

modelo_rn = Sequential()

modelo_rn.add(Dense(32, activation='relu', input_shape=(X_treino.shape[1],)))
modelo_rn.add(Dense(16, activation='relu'))

# 6 classes
modelo_rn.add(Dense(6, activation='softmax'))

modelo_rn.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Iniciando o treinamento da rede neural...\n") #\n no final quebrar linha

historico = modelo_rn.fit( X_treino,  y_treino, epochs=100,verbose=0 )

print("Treinamento concluído!")

perda, acuracia = modelo_rn.evaluate(X_teste, y_teste)

print(f"\nAcurácia na prova final: {acuracia * 100:.2f}%")

```

```python
# Célula 7: Simulando a Análise de um novo cliente

# 1. Entrada de dados de um cliente solicitando crédito hoje
novo_cliente = {
    'Idade': [22],
    'Salario': [3000],
    'Divida_Atual': [900]
}

novo_client_df = pd.DataFrame(novo_cliente)

#2. Passo essencial : Passar o cliente pelo mesmo padronizador matemático
novo_cliente_normalizado = normalizado.transform(novo_client_df)

#3 . Pedindo para a Rede Neural calcular a probabilidade de risco
predicao = modelo_rn.predict(novo_cliente_normalizado)
probabilidade_risco = predicao[0][0]

print(f"Relatório de Análise:{probabilidade_risco *100: 2f}% de probalidade de inadimplencia.\n")

# 4 Regra de Négocio da Empresa (Ponto de corte de 50%)
if probabilidade_risco > 0.5:
    print("CRÉDITO REPROVADO : Identificado na análise alto risco de inadimplência")
else:
    print("CRÉDITO APROVADO: O perfil encontra-se dentro da margem de segurança ")
```
