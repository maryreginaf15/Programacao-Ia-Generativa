```python
# Conhecendo o Pandas

import pandas as pd

# 1. Criando um dicionario
dados_exemplo = {
    "Nome": ["Ana", "João", "Mariana"],
    "Idade": [25, 30, 22],
    "Profissão": ["Engenheira", "Médico", "Analista de Dados"]
}

#2. Transformando o dicionario em uma tabela (DataFrame)
df = pd.DataFrame(dados_exemplo)

print("Visualizando os dados organizados com o Pandas")
display(df)

#3. Separando apenas a coluna de Idades
idades = df["Idade"]
print("\nApenas a coluna de Idades")
display(idades)
```
